```python
import random
import numpy as np
import torch
from skimage import data
from skimage.transform import resize

im = data.coffee()
im = resize(im, (im.shape[0] // 8, im.shape[1] // 8), mode='reflect', preserve_range=True, anti_aliasing=True).astype(np.uint8)
im = np.swapaxes(np.swapaxes(im, 0, 2), 1, 2)
im = torch.from_numpy(im)

a = random.uniform(0,2)
b = random.uniform(-50,50)
```

```python
im.shape
```

```python
im.dtype
```

```python
import matplotlib.pyplot as plt

img = im.swapaxes(0, 2)
img = img.swapaxes(0, 1)

plt.imshow(img, cmap='gray')
plt.show()
```

```python
out = (a * im + b)

out = torch.round(out)
out = torch.clip(out, min=0, max=255).to(torch.uint8)
```

```python
out = out.swapaxes(0, 2)
out = out.swapaxes(0, 1)
```
